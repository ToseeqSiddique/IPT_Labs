Task 02:write a program that accepts an image's width and height and calculates its total number of pixels.

```python
width = int(input("Enter image width: "))
height = int(input("Enter image height: "))

pixels = width * height

print("Total Pixels =", pixels)
```

Task 03: 3.	Write a program that accepts an image’s width and height and calculates its uncompressed grayscale size in kilobytes.

```python
width = int(input("Enter image width: "))
height = int(input("Enter image height: "))

size = (width * height) / 1024

print("Grayscale Image Size =", size, "KB")

```

Task 04:Modify Program 1 to accept a student’s name, roll number, and marks in three subjects, then calculate the total and average marks.

```python
name = input("Enter student name: ")
roll = input("Enter roll number: ")

subject1 = float(input("Enter marks in Subject 1: "))
subject2 = float(input("Enter marks in Subject 2: "))
subject3 = float(input("Enter marks in Subject 3: "))

total = subject1 + subject2 + subject3
average = total / 3

print("\nStudent Name:", name)
print("Roll Number:", roll)
print("Total Marks:", total)
print("Average Marks:", average)
```
