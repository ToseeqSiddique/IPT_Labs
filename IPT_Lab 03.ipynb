Task 01:	Change the numbers array to [5, 10, 15] and calculate its sum and average.

```python
numbers = [5, 10, 15]

total = sum(numbers)
average = total / len(numbers)

print("Array =", numbers)
print("Sum =", total)
print("Average =", average)
```

Task 02:Change the text in the OpenCV program to your own message.

```python
import cv2
import numpy as np

img = np.zeros((300, 400, 3), dtype=np.uint8)

cv2.putText(img, "Welcome to IPT", (50, 150),
            cv2.FONT_HERSHEY_SIMPLEX, 1,
            (255, 255, 255), 2)

cv2.imshow("My Message", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Task 03: Change image dimensions from (300, 400, 3) to (200, 500, 3).

```python
import cv2
import numpy as np

img = np.zeros((200, 500, 3), dtype=np.uint8)

cv2.imshow("New Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()

print("Image dimensions:", img.shape)
```

Task 04: Try a red text colour in OpenCV: (0, 0, 255) because OpenCV uses BGR order.

```python
import cv2
import numpy as np

img = np.zeros((300, 400, 3), dtype=np.uint8)

cv2.putText(img, "Red Text", (50, 150),
            cv2.FONT_HERSHEY_SIMPLEX, 1,
            (0, 0, 255), 2)

cv2.imshow("Red Text", img)
cv2.waitKey(0)
cv2.destroyAllWindows()

```
